#指数基金分类图

```python
import matplotlib.pyplot as plt
import matplotlib

# 设置中文字体（解决Jupyter绘图中文乱码问题）
# 如果是Windows系统，通常使用黑体 'SimHei'
#matplotlib.rcParams['font.sans-serif'] = ['SimHei'] 
#matplotlib.rcParams['axes.unicode_minus'] = False 

def draw_index_taxonomy():
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # 定义节点坐标
    nodes = {
        '指数基金': (5, 5),
        '宽基指数': (2, 3),
        '窄基(行业)指数': (4, 3),
        '策略(Smart Beta)': (6, 3),
        '跨境指数': (8, 3),
        '沪深300/中证500': (2, 1),
        '消费/医疗/半导体': (4, 1),
        '红利/低波/价值': (6, 1),
        '纳指/标普/恒生': (8, 1)
    }
    
    # 绘制连线
    connections = [
        ('指数基金', '宽基指数'), ('指数基金', '窄基(行业)指数'),
        ('指数基金', '策略(Smart Beta)'), ('指数基金', '跨境指数'),
        ('宽基指数', '沪深300/中证500'), ('窄基(行业)指数', '消费/医疗/半导体'),
        ('策略(Smart Beta)', '红利/低波/价值'), ('跨境指数', '纳指/标普/恒生')
    ]
    
    for start, end in connections:
        x_values = [nodes[start][0], nodes[end][0]]
        y_values = [nodes[start][1], nodes[end][1]]
        ax.plot(x_values, y_values, c='gray', linestyle='--', zorder=1)

    # 绘制节点
    for label, (x, y) in nodes.items():
        color = 'skyblue' if y == 5 else ('lightgreen' if y == 3 else 'orange')
        ax.scatter(x, y, s=3000, c=color, edgecolors='black', zorder=2)
        ax.text(x, y, label, ha='center', va='center', fontsize=10, fontweight='bold')

    # 设置图形属性
    ax.set_title('指数基金分类脑图', fontsize=16, pad=20)
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 6)
    ax.axis('off')

    plt.show()

draw_index_taxonomy()
```
